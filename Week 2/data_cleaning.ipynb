```python
import pandas as pd

sales = pd.read_csv("../Week 1/sales_result.csv")

sales.head()
```

```python
sales.isnull().sum()
```

```python
sales["discount"] = sales["discount"].fillna(sales["discount"].mean())

sales
```

```python
sales.duplicated().sum()
```

```python
Q1 = sales["price"].quantile(0.25)
Q3 = sales["price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

sales[(sales["price"] < lower) | (sales["price"] > upper)]
```

```python
sales["price"] = sales["price"].clip(lower=lower, upper=upper)

sales.head()
```

```python
text_columns = ["customer_name", "product", "category", "city"]

for col in text_columns:
    sales[col] = sales[col].str.strip().str.title()

sales.head()
```

```python
sales.isnull().sum()
```

```python
print("AFTER CLEANING")
print("Missing values:", sales.isnull().sum().sum())
print("Duplicate rows:", sales.duplicated().sum())
print("Total rows:", len(sales))
```

```python
sales.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
