# Week 2 Day 5 — NumPy for Data

This notebook demonstrates NumPy arrays, vectorised operations,
boolean masking, statistical analysis, and performance comparison.

```python
import numpy as np

# 1D array
array_1d = np.array([10, 20, 30, 40, 50])

# 2D array
array_2d = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

# 3D array
array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Array:")
print(array_1d)

print("\n2D Array:")
print(array_2d)

print("\n3D Array:")
print(array_3d)
```

```python
print("1D shape:", array_1d.shape)
print("2D shape:", array_2d.shape)
print("3D shape:", array_3d.shape)
```

```python
sales = np.array([100, 200, 300, 400, 500])

# Addition
print("Add 50:", sales + 50)

# Multiplication
print("Double sales:", sales * 2)

# Total
print("Total sales:", np.sum(sales))

# Mean
print("Average sales:", np.mean(sales))

# Maximum
print("Maximum sales:", np.max(sales))

# Minimum
print("Minimum sales:", np.min(sales))
```

```python
regional_sales = np.array([
    [100, 200, 300],
    [150, 250, 350],
    [120, 220, 320]
])

print("Total:", np.sum(regional_sales))

print("Column totals:", np.sum(regional_sales, axis=0))

print("Row totals:", np.sum(regional_sales, axis=1))

print("Column averages:", np.mean(regional_sales, axis=0))

print("Row averages:", np.mean(regional_sales, axis=1))
```

```python
sales = np.array([100, 250, 150, 500, 350, 90, 700])

# Select sales greater than 300
high_sales = sales[sales > 300]

print("Sales greater than 300:")
print(high_sales)
```

```python
sales = np.array([100, 200, 300, 400, 500])

selected_sales = sales[[0, 2, 4]]

print("Selected values:", selected_sales)
```

```python
sales = np.array([100, 200, 300, 400, 500, 600, 700])

mean_sales = np.mean(sales)
std_sales = np.std(sales)
percentile_50 = np.percentile(sales, 50)

print("Mean:", mean_sales)
print("Standard deviation:", std_sales)
print("50th percentile:", percentile_50)
```

```python
advertising = np.array([10, 20, 30, 40, 50])
sales = np.array([100, 180, 250, 350, 450])

correlation_matrix = np.corrcoef(advertising, sales)

print("Correlation matrix:")
print(correlation_matrix)

print("Correlation coefficient:", correlation_matrix[0, 1])
```

```python
import time

data = np.random.rand(1_000_000)
```

```python
start = time.time()

loop_result = []

for value in data:
    loop_result.append(value * 2)

loop_time = time.time() - start

print("Python loop time:", loop_time)
```

```python
start = time.time()

numpy_result = data * 2

numpy_time = time.time() - start

print("NumPy vectorised time:", numpy_time)
```

```python
print("Python loop:", loop_time)
print("NumPy:", numpy_time)

print("NumPy is approximately",
      round(loop_time / numpy_time, 2),
      "times faster.")
```

## Conclusion

NumPy provides efficient numerical computation using arrays and vectorised operations.
I used 1D, 2D and 3D arrays, boolean masking, fancy indexing and aggregation across axes.
I also calculated statistical measures such as mean, standard deviation, percentile and correlation.

The performance comparison showed that NumPy vectorised operations are faster than Python loops
for large numerical datasets. This makes NumPy useful for efficient data analysis pipelines.
