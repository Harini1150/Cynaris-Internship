## Hypothesis Scenarios

### Scenario 1 — Average Price
Business claim: The average product price is ₹10,000.

- Null hypothesis (H₀): The population mean price is ₹10,000.
- Alternative hypothesis (H₁): The population mean price is not ₹10,000.

### Scenario 2 — Discount
Business claim: The average discount is 10%.

- Null hypothesis (H₀): The population mean discount is 10%.
- Alternative hypothesis (H₁): The population mean discount is not 10%.

### Scenario 3 — Quantity Comparison
Business question: Is the average quantity sold different between Electronics and Clothing?

- Null hypothesis (H₀): The mean quantity sold for Electronics equals the mean quantity sold for Clothing.
- Alternative hypothesis (H₁): The mean quantity sold for Electronics is different from the mean quantity sold for Clothing.

```python
price_data = df["price"]

t_stat, p_value = stats.ttest_1samp(price_data, popmean=10000)

print("One-Sample t-Test")
print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
confidence_level = 0.95
mean_price = price_data.mean()
std_error = stats.sem(price_data)
df_degrees = len(price_data) - 1

confidence_interval = stats.t.interval(
    confidence_level,
    df=df_degrees,
    loc=mean_price,
    scale=std_error
)

print("Mean price:", mean_price)
print("95% Confidence Interval:", confidence_interval)
```

```python
print("One-Sample t-Test")
print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

```python
electronics_quantity = df[
    df["category"] == "Electronics"
]["quantity"]

clothing_quantity = df[
    df["category"] == "Clothing"
]["quantity"]

print("Electronics quantities:")
print(electronics_quantity.values)

print("\nClothing quantities:")
print(clothing_quantity.values)
```

```python
t_stat_2, p_value_2 = stats.ttest_ind(
    electronics_quantity,
    clothing_quantity,
    equal_var=False
)

print("Two-Sample t-Test")
print("t-statistic:", t_stat_2)
print("p-value:", p_value_2)

if p_value_2 < 0.05:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

### Two-Sample t-Test Interpretation

The mean quantity for Electronics was 1.0, while the mean quantity for Clothing was 2.5.

The Welch two-sample t-test produced a t-statistic of -3.0 and a p-value of 0.2048. Since the p-value is greater than 0.05, we fail to reject the null hypothesis.

There is insufficient statistical evidence to conclude that the average quantity sold differs between Electronics and Clothing.

The result should be interpreted cautiously because the sample sizes are very small and the Electronics group has no variation in quantity. SciPy also reported a precision-loss warning due to the nearly identical values in the Electronics group.

## Type I and Type II Errors in Business

### Type I Error (False Positive)
A Type I error occurs when we reject the null hypothesis even though it is actually true.

**Business example:**  
A company concludes that a new website design increases sales when in reality it does not. The company may spend additional resources implementing a change that provides no real benefit.

### Type II Error (False Negative)
A Type II error occurs when we fail to reject the null hypothesis even though the alternative hypothesis is actually true.

**Business example:**  
A company concludes that a new website design does not improve sales when it actually does. The company may miss an opportunity to improve customer conversions.

### Key Difference

- **Type I:** False positive — detecting an effect that does not actually exist.
- **Type II:** False negative — failing to detect an effect that actually exists.
