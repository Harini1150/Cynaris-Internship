```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Week 2/cleaned_dataset.csv")

df.head()
```

```python
df.info()
```

```python
# Select numerical columns
numerical_columns = ["quantity", "price", "discount"]

# Calculate descriptive statistics
statistics = df[numerical_columns].describe()

statistics
```

```python
# Calculate mode
mode_values = df[["quantity", "price", "discount"]].mode()

# Calculate variance
variance_values = df[["quantity", "price", "discount"]].var()

# Calculate standard deviation
std_values = df[["quantity", "price", "discount"]].std()

print("Mode:")
print(mode_values)

print("\nVariance:")
print(variance_values)

print("\nStandard Deviation:")
print(std_values)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["price"], bins=5, edgecolor="black")

plt.title("Price Frequency Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

```python
# Calculate skewness
skewness = df["price"].skew()

print("Price Skewness:", skewness)

if skewness > 0:
    print("Distribution is positively skewed (right-skewed)")
elif skewness < 0:
    print("Distribution is negatively skewed (left-skewed)")
else:
    print("Distribution is approximately symmetric")
```

```python
# Calculate percentiles
percentiles = df["price"].quantile([0.25, 0.50, 0.75, 0.90])

print("Price Percentiles:")
print(percentiles)
```

```python
df["category"].value_counts()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="category",
    y="price"
)

plt.title("Price Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Price")

plt.show()
```

```python
# Statistical Summary

print("STATISTICAL SUMMARY")
print("===================")

print("\nMean:")
print(df[["quantity", "price", "discount"]].mean())

print("\nMedian:")
print(df[["quantity", "price", "discount"]].median())

print("\nMode:")
print(df[["quantity", "price", "discount"]].mode().iloc[0])

print("\nVariance:")
print(df[["quantity", "price", "discount"]].var())

print("\nStandard Deviation:")
print(df[["quantity", "price", "discount"]].std())
```
