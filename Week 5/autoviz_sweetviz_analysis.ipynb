## 3 Key Insights from Automated EDA

1. **Category performance:** Electronics generated the highest total sales amount, followed by Clothing and Furniture.

2. **Profit performance:** Clothing generated the highest total profit, while Furniture contributed comparatively lower profit.

3. **Sales and profit relationship:** The automated EDA reports show the relationship between Amount and Profit, helping identify how sales value is associated with profitability.

> Note: These insights should be validated against the underlying dataset before making business decisions.

### Insight 1: Sales, Quantity and Profit Relationships

The AutoViz correlation heatmap shows a weak positive relationship between Amount and Quantity (correlation = 0.35). Amount and Profit also have a weak positive relationship (0.24), while Profit and Quantity have almost no linear relationship. This indicates that higher sales amounts are somewhat associated with larger quantities and higher profit, but the relationships are not strong enough to suggest that one variable alone determines another.

### Insight 2: Quantity and Average Order Amount

The AutoViz categorical analysis shows that average order Amount generally increases at higher Quantity levels, although the relationship is not perfectly consistent across all quantities. Orders with quantities of 13 and 14 have particularly high average Amount values. This suggests that higher-quantity orders can contribute substantially to sales value.

### Insight 3: Distribution of Order Amount

The AutoViz KDE distribution plot shows that Order Amount has a two-peaked (bimodal) distribution. Most observations are concentrated around the lower Amount range, while a smaller secondary concentration appears around approximately 1,000–1,500. This indicates that the dataset contains both lower-value and higher-value order groups rather than a single consistent order-value pattern.
## AutoViz vs SweetViz Comparison

AutoViz quickly generates a broad range of visualizations with minimal code, making it useful for rapid exploratory analysis. SweetViz produces a structured HTML report that organizes variable-level summaries, distributions, associations, and data-quality information. AutoViz is more visualization-oriented, while SweetViz provides a more organized report that is easier to navigate and share. Both tools reduce repetitive EDA coding and help analysts identify patterns quickly. However, automated reports still require analyst validation and business context. For business stakeholders, SweetViz is suitable as a primary report because its structured HTML format is easier to review and share, while AutoViz can support exploratory analysis during the analyst's investigation.

## Recommendation for Business Stakeholders

SweetViz is recommended as the primary automated EDA reporting tool for business stakeholders because it provides a structured HTML report that is easier to review and share. AutoViz is useful for analysts during exploratory analysis because it generates multiple visualizations quickly. A practical workflow is to use AutoViz for initial exploration and pattern discovery, then use SweetViz to present the key findings in a more organized format. Regardless of the tool used, automated findings should be validated against the underlying data before making business decisions.
