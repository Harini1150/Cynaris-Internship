```python
import logging

logging.basicConfig(
    filename="pandasai_errors.log",
    level=logging.ERROR,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def safe_pandasai_query(question):
    try:
        result = agent.chat(question)

        result_text = str(result)

        if "Error:" in result_text or "error" in result_text.lower():
            logging.error(
                "PandasAI query failed: %s | Result: %s",
                question,
                result_text
            )
            print("Query failed.")
            print(result)
            return None

        print("Query successful!")
        print(result)
        return result

    except Exception as e:
        logging.error(
            "PandasAI query failed: %s | Exception: %s",
            question,
            str(e)
        )
        print("Query failed.")
        print("Error:", e)
        return None
```

```python
response = safe_pandasai_query(
    "Give me the total Amount for a column_that_does_not_exist"
)
```

```python
response = safe_pandasai_query(
    "Calculate the average of the column XYZ_INVALID_COLUMN_999"
)
```
